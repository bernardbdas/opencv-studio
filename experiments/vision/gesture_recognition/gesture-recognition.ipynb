# MediaPipe Gesture Recognizer Exploration
This notebook demonstrates how to load the MediaPipe Gesture Recognizer model, process static images, and extract recognized gestures and landmarks.

```python
import os
import cv2
import mediapipe as mp
import numpy as np
import requests

# Download model if not exists
model_path = "../models/gesture_recognizer/gesture_recognizer.task"
os.makedirs(os.path.dirname(model_path), exist_ok=True)
if not os.path.exists(model_path):
    url = "https://storage.googleapis.com/mediapipe-models/gesture_recognizer/gesture_recognizer/float16/latest/gesture_recognizer.task"
    print("Downloading gesture recognition model...")
    r = requests.get(url)
    with open(model_path, "wb") as f:
        f.write(r.content)
    print("Download complete!")

from mediapipe.tasks.python.core.base_options import BaseOptions
from mediapipe.tasks.python.vision import RunningMode, GestureRecognizer, GestureRecognizerOptions

base_options = BaseOptions(model_asset_path=model_path)
options = GestureRecognizerOptions(
    base_options=base_options,
    running_mode=RunningMode.IMAGE,
    num_hands=2,
)
recognizer = GestureRecognizer.create_from_options(options)
print("Gesture Recognizer initialized successfully!")
```
